Q1. Write a program to find the following between two vectors a = [2,3,7,8,7] and b = [4,7,8,2,1].
- Euclidean distance
- Manhattan distance
- Cosine Similarity

```python
import numpy as np

a= np.array([2,3,7,8,7])
b= np.array([4,7,8,2,1])

e_d = np.sqrt(np.sum((a-b)**2))
print(f"Euclidean Distance: {e_d}")

m_d = np.sum(np.abs(a-b))
print(f"Manhattan Distance: {m_d}")

dot = np.dot(a,b)
norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)
c_s= dot / (norm_a * norm_b)
print(f"Cosine Similarity: {c_s}")

```

Q2. Write a program to find a simple matching coefficient and Jaccard similarity between [1,0,1,0,0,0] and [1,0,1,1,0,1].

```python
v1 = np.array([1, 0, 1, 0, 0, 0])
v2 = np.array([1, 0, 1, 1, 0, 1])

# counting the matches between the vectors v1 and v2 
f11 =np.sum((v1==1) & (v2==1))  # both are 1
f00 = np.sum((v1==0) & (v2==0))  # both are 0
f10 = np.sum((v1==1) & (v2==0))  # v1 is 1 and v2 is 0
f01 = np.sum((v1==0) & (v2==1))  # v1 is 0 and v2 is 1

# simple matching coefficient
s_m_c = (f11 + f00) / (f11 + f00 + f10 + f01)
print(f"Simple Matching Coefficient: {s_m_c}")

# jaccard coefficient
j_c = f11 / (f11 + f10 + f01)
print(f"Jaccard Coefficient: {j_c}")
```

Q3. Write a program to perform feature scaling (min max, z score and decimal scaling) on [35000, 42000, 39000, 45000]
[1-3 using numpy only]

```python
data = np.array([35000, 42000, 39000, 45000], dtype=float)

# feature scaling
# 1. min-max scaling 

min_val = np.min(data)
max_val = np.max(data)
min_max_scaled = (data - min_val) / (max_val - min_val)
print(f"Min-Max Scaled Data: {min_max_scaled}")

# 2. z score scaling 
mean = np.mean(data)
std_dev = np.std(data)
z_score_scaled = (data - mean) / std_dev
print(f"Z-Score Scaled Data: {z_score_scaled}")

# 3. decimal scaling
max_d = np.max(np.abs(data))
decimal_scaled = data / (10 ** np.ceil(np.log10(max_d)))
print(f"Decimal Scaled Data: {decimal_scaled}")
```

Q4. You are provided with the following arrays representing the monthly sales data (in USD) of two products — Laptops and Smartphones — sold by an electronics store over the first 6 months of the year:
months = ['January', 'February', 'March', 'April', 'May', 'June']
laptop_sales = [12000, 15000, 17000, 13000, 16000, 18000]
smartphone_sales = [10000, 14000, 16000, 12000, 15500, 17500]

- Plot the sales trends of both laptops and smartphones over the 6 months.
- Create a scatter plot comparing laptop sales vs smartphone sales.
- Draw a bar chart showing sales of laptops and smartphones for each month.
- Plot a pie chart showing the percentage contribution of each month to total
laptop sales.
- Plot another pie chart for smartphone sales.

```python
import matplotlib.pyplot as plt

months = ['January', 'February', 'March', 'April', 'May', 'June']
laptop_sales = [12000, 15000, 17000, 13000, 16000, 18000]
smartphone_sales = [10000, 14000, 16000, 12000, 15500, 17500]

# plotting the sales trends over 6 months

plt.figure(figsize=(10, 5))
plt.plot(months, laptop_sales, marker='o', label='Laptop Sales',color='pink')
plt.plot(months, smartphone_sales, marker='o', label='Smartphone Sales',color='brown')
plt.title('Sales Trends Over 6 Months')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.legend()
plt.grid(True)
plt.show()

# scatter plot comparing laptop and smartphone sales
plt.figure(figsize=(8, 6))
plt.scatter(laptop_sales, smartphone_sales, color='purple')
plt.title('Laptop vs Smartphone Sales') 
plt.xlabel('Laptop Sales')
plt.ylabel('Smartphone Sales')
plt.grid(True)
plt.show()

# bar chart showing sales for each month
plt.figure(figsize=(10, 5))
x = np.arange(len(months))
plt.bar(x, laptop_sales, label='Laptop Sales', color='pink')
plt.bar(x, smartphone_sales, label='Smartphone Sales', color='brown')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.title('Sales for Each Month')
plt.xticks(x, months)
plt.legend()
plt.grid(True)
plt.show()

# pie chart showing the percentage contribution of each month to total laptop sales
plt.figure(figsize=(8, 8))
plt.pie(laptop_sales, labels=months, autopct='%1.1f%%', startangle=140, colors=['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0','#ffb3e6'])
plt.title('Percentage Contribution of Each Month to Total Laptop Sales')
plt.axis('equal') 
plt.show()

# pie chart for smartphone sales
plt.figure(figsize=(8, 8))
plt.pie(smartphone_sales, labels=months, autopct='%1.1f%%', startangle=140, colors=['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0','#ffb3e6'])
plt.title('Percentage Contribution of Each Month to Total Smartphone Sales')
plt.axis('equal')
plt.show()

```
