Q1. Write a program that takes three numbers and prints the largest one using
if...elif...else

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
num3 = float(input("Enter third number: "))

if (num1 >= num2) and (num1 >= num3):
    largest = num1
elif (num2 >= num1) and (num2 >= num3):
    largest = num2
else:
    largest = num3

print(f"The largest number is: {largest}")
```

Q2. Print the multiplication table of a given number using a for loop.


```python
num = int (input("Enter a number for the multiplication table: "))
print(f"Multiplication table for {num}:")
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

Q3. Write a program that keeps asking the user to enter a number until they enter -1, and then prints the sum of all entered numbers (excluding -1).


```python
sum = 0
print("Enter numbers to add. Enter -1 to stop.")

while True:
    num = float(input("Enter a number: "))
    if num == -1:
        break
    sum += num

print(f"The sum of the entered numbers is: {sum}")
```

Q5. Write a function is_prime(n) that returns True if a number is prime, else returns False.

```python
def is_prime(n):
    if n<=1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False
    return True

num = int(input("Enter a number to check if it is prime: "))
if is_prime(num):
    print(f"{num} is a prime number.")

else:
    print(f"{num} is not a prime number.")
```

Q6. Write a function is_palindrome(word) that returns True if the word is a palindrome(same forwards and backwards), and False otherwise.

```python
def is_palindrome(word):
    word = word.lower()

    re_word = word[::-1]

    if word == re_word:
        print(f"{word} is a palindrome.")
    else:
        print(f"{word} is not a palindrome.")

word= input("Enter a word to check if it is a palindrome: ")
is_palindrome(word)
```

Q7. Create a Python file named calculator.py that contains the following functions:
- add(a, b)
- subtract(a, b)
- multiply(a, b)
- divide(a, b).

Then, write another Python script that imports this module and calls each function with sample inputs.

```python
%%writefile calculator.py

# this created the another file as calculator.py in the same place where this ipynb file is located.

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        raise ValueError("Cannot divide by zero.")
    return a / b

```

```python
# now importing the same calulator.py file to use the functions defined there

import calculator as calc

x= int(input("Enter first number: "))
y= int(input("Enter second number: "))

print(f"Addition: {calc.add(x, y)}")
print(f"Subtraction: {calc.subtract(x, y)}")
print(f"Multiplication: {calc.multiply(x, y)}")
print(f"Division: {calc.divide(x, y)}")
```
