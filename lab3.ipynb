```python
# e-commerce data operation

import pandas as pd 
import numpy as np

# loading the dataset as per the question

ecommerce_data = {
    'OrderID': [101, 102, 103, 104, 105, 106, 107],
    'CustomerName': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace'],
    'Category': ['Electronics', 'Fashion', 'Electronics', 'Home', 'Fashion', 'Electronics', 'Home'],
    'Amount': [200, 150, 300, 400, 250, 500, 350],
    'Discount': [20, 10, 30, 50, 20, 60, 40],
    'PurchaseDate': pd.to_datetime(['2023-11-20', '2023-11-21', '2023-11-22', '2023-11-23', '2023-11-24', '2023-11-25', '2023-11-26']),
    'City': ['New York', 'Los Angeles', 'New York', 'Chicago', 'Los Angeles', 'Chicago', 'New York']
}

df_ecommerce = pd.DataFrame(ecommerce_data)

# 1. Shape of dataset
print("1. Shape of the dataset:", df_ecommerce.shape, "\n")

# 2. Are any null values? 
print("2. Null values in the dataset:")
print(df_ecommerce.isnull().sum(), "\n")
print("Total null values in the dataset:", df_ecommerce.isnull().sum().sum())

# 3. Retrieve [Charlie, Electronics, 300] using loc and iloc

print("\n3. Retrieve [Charlie, Electronics, 300] using loc and iloc:")
print("\n Using loc:")
print(df_ecommerce.loc[df_ecommerce['CustomerName'] == 'Charlie', ['CustomerName', 'Category', 'Amount']])
print("\n Using iloc:")
print(df_ecommerce.iloc[2])

''' 
The difference is that loc selects data by labels
i.e. names as in this case 'Charlie' and the specific columns,
while iloc selects data by integer index positions. 
In this case, iloc[2] retrieves the third row of the DataFrame, 
which corresponds to Charlie's data.

'''
```

```python
# 4. Fetch rows for orders placed by customers from New York  

print("\n4. Fetch rows for orders placed by customers from New York:\n")
ny_orders = df_ecommerce[df_ecommerce['City'] == 'New York']
print(ny_orders)

# 5. What is the average order amount? What is the total discount provided?  

avg_order_amount = df_ecommerce['Amount'].mean()
total_discount = df_ecommerce['Discount'].sum()

print(f"\n5. Average order amount and the discount provided are:  \n\n Average order amount: {avg_order_amount}")
print(f" Total discount provided: {total_discount}")
```

```python
# 6. Sort the orders by Amount in descending order  

print("\n6. Sort the orders by Amount in descending order:\n")
sorted_orders = df_ecommerce.sort_values(by='Amount', ascending=False)
print(sorted_orders)
```

```python
# 7. Display all orders where the Category is "Electronics" and the Amount is greater than 250  

print("\n7. Display all orders where the Category is 'Electronics' and the Amount is greater than 250:\n")
electronics_orders = df_ecommerce[(df_ecommerce['Category'] == 'Electronics') & (df_ecommerce['Amount'] > 250)]
print(electronics_orders)
```

# Data Preprocessing in the second dataset
---

```python
# loading the dataset for preprocessing

data = [
    {"id": 1, "age": 25, "income": 50000, "city": "New York", "purchased": "yes"},
    {"id": 2, "age": 30, "income": 60000, "city": "San Francisco", "purchased": "no"},
    {"id": 3, "age": None, "income": 75000, "city": "New York", "purchased": None},
    {"id": 4, "age": 22, "income": None, "city": "Los Angeles", "purchased": "yes"},
    {"id": 5, "age": 35, "income": 80000, "city": None, "purchased": "no"},
    {"id": 6, "age": 29, "income": 58000, "city": "San Francisco", "purchased": "yes"},
    {"id": 7, "age": None, "income": None, "city": "New York", "purchased": "no"},
    {"id": 8, "age": 40, "income": 90000, "city": "San Francisco", "purchased": "yes"},
    {"id": 9, "age": 28, "income": 65000, "city": "Los Angeles", "purchased": None},
    {"id": 10, "age": 33, "income": 72000, "city": "New York", "purchased": "no"},
    {"id": 11, "age": None, "income": 50000, "city": "Los Angeles", "purchased": "yes"},
    {"id": 12, "age": 26, "income": 61000, "city": None, "purchased": "yes"},
    {"id": 13, "age": 45, "income": 95000, "city": "San Francisco", "purchased": None},
    {"id": 14, "age": 50, "income": None, "city": "Los Angeles", "purchased": "no"},
    {"id": 15, "age": 38, "income": 70000, "city": "New York", "purchased": "yes"}
]

df = pd.DataFrame(data)
print(df.head())
```

```python
# 1. Replace None in numerical fields (age, income) with median/mean[cite: 1]

df['age'] = df['age'].fillna(df['age'].median())
df['income'] = df['income'].fillna(df['income'].mean())

# 2. Replace None in categorical fields (city, purchased) with placeholder "unknown"

df['city'] = df['city'].fillna("unknown")
df['purchased'] = df['purchased'].fillna("unknown")

# 3. Normalize numerical fields (age, income) using Min-Max scaling

df['norm_age'] = (df['age'] - df['age'].min()) / (df['age'].max() - df['age'].min())
df['norm_income'] = (df['income'] - df['income'].min()) / (df['income'].max() - df['income'].min())

# 4. Use one-hot encoding for the city

df = pd.get_dummies(df, columns=['city'], prefix='city')

# 5. Use label encoding for the purchase i.e. yes =1, no = 0 and unknown = -1

pur_mapping = {'yes': 1, 'no': 0, 'unknown': -1}
df['purchased'] = df['purchased'].map(pur_mapping)

# Processed dataframe head

print("\nProcessed DataFrame head:\n")
print(df.head())
```

# Reflection of the processing

> Here the missing numerical values of age and income were imputed on the basis of mean, i.e. in id 3 of age and id 4 of income. 

> The missing city was replaced with unknown so in the id 4 before preprocessing there was Nan in city which is later replaced by unknown in the processed one. 

> Age and income normalized within the range [0,1] as shown as two different columns in processed dataframe. 

> One hot encoding coverted categorical city variable into binary columns as True or False and label encoding mapped the target purchased into the numerical categories. 

Thus, doing this our unprocessed data is now fully compatible with machine learning algorithms.
